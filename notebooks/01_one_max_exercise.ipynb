```python
# ============================================================
# Main Evolutionary Loop (Simplified)
# ============================================================

"""
In this section, we evolve our population step by step.

Each generation:
1. Select parents (based on fitness)
2. Apply crossover and mutation to create offspring
3. Evaluate new individuals
4. Replace the old population
5. Record statistics
"""

### TODO: initialize generation counter
generationCounter = 0  # already done for you

# Stop if max fitness == target or if max generations reached
while max(fitnessValues) < ONE_MAX_LENGTH and generationCounter < MAX_GENERATIONS:

    # Step 1: Increase generation count
    generationCounter += 1

    # Step 2: Select and clone individuals
    ### TODO: use toolbox.select() and map toolbox.clone()
    # offspring = toolbox.select(...)

    offspring = None # TODO
    offspring = list(map(toolbox.clone, offspring))

    # Step 3: Apply crossover
    for child1, child2 in zip(offspring[::2], offspring[1::2]):
        if random.random() < P_CROSSOVER:
            ### TODO: apply crossover with mate

            # Delete fitness (so it will be re-evaluated)
            del child1.fitness.values
            del child2.fitness.values

    # Step 4: Apply mutation
    for mutant in offspring:
        if random.random() < P_MUTATION:
            ### TODO: apply mutation with mutate

            del mutant.fitness.values

    # Step 5: Evaluate new individuals (only those with no fitness)
    freshIndividuals = [ind for ind in offspring if not ind.fitness.valid]
    freshFitnessValues = list(map(toolbox.evaluate, freshIndividuals))
    for ind, fitVal in zip(freshIndividuals, freshFitnessValues):
        ind.fitness.values = fitVal

    # Step 6: Replace the old population
    population = offspring

    # Step 7: Gather and record statistics
    fitnessValues = [ind.fitness.values[0] for ind in population]
    maxFitness = max(fitnessValues)
    meanFitness = sum(fitnessValues) / len(population)
    maxFitnessValues.append(maxFitness)
    meanFitnessValues.append(meanFitness)

    # Print generation summary
    print(f"Generation {generationCounter}: Max Fitness = {maxFitness}, Avg Fitness = {meanFitness:.2f}")
```

```python
# ============================================================
# Visualize the Results
# ============================================================

# Plot the max and average fitness values over generations
sns.set_style("whitegrid")
plt.plot(maxFitnessValues, color='red', label='Max Fitness')
plt.plot(meanFitnessValues, color='green', label='Avg Fitness')
plt.xlabel('Generation')
plt.ylabel('Fitness')
plt.title('OneMax: Evolution of Fitness Over Generations')
plt.legend()
plt.show()
```
